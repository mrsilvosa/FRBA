# Ejercicio 1:
## Transferencia de tensión
### Cuadripolo LC cargado con Rg en la entrada, fuente de tensión Vg

$$
T(s) = \frac{V_2}{V_g} = K \cdot \frac{s \left(s^2 + \frac{1}{16}\right)}{s^3 + 2 s^2 + 2 s + 1}
$$

$$
Denominador: \ \ s^3 + 2 s^2 + 2 s + 1 = (s+1)\left(s + \frac{1}{2} - j\frac{\sqrt{3}}{2}\right)\left(s + \frac{1}{2} + j\frac{\sqrt{3}}{2}\right) = (s+1)\left(s^2 + s + 1\right)
$$

### Entrada de corriente reestringida por Rg, salida del circuito abierta, me conviene pensar al cuadripolo como $Z$:

$$
\begin{cases}
V_1 = Z_{11} \cdot I_1 + Z_{12} \cdot I_2 \\
V_2 = Z_{21} \cdot I_1 + Z_{22} \cdot I_2
\end{cases}
$$

$$
Z_{11} = \left. \frac{V_1}{I_1} \right|_{I_2=0} \quad ; \quad Z_{12} = \left. \frac{V_1}{I_2} \right|_{I_1=0}
$$

$$
Z_{21} = \left. \frac{V_2}{I_1} \right|_{I_2=0} \quad ; \quad Z_{22} = \left. \frac{V_2}{I_2} \right|_{I_1=0}
$$

#### $V_1$ será la tensión entre los bornes de la entrada del cuadripolo, distinto de $V_g$:

### Si reducimos la expresión a la entrada:
$$
Z_{11} = \left. \frac{V_1}{I_1} \right|_{I_2=0}
$$
Cuadripolo reducido a $Z_{11}$ (salida en circuito abierto, $I_2=0$):

$$
\frac{V_1}{V_g} = \frac{Z_{11}}{Z_{11} + R_g} \quad ;
$$

$$
\left. \frac{V_2}{V_g} \right|_{I_2=0} = \frac{V_2}{I_1} \cdot \left(\frac{V_1}{I_1}\right)^{-1} \cdot \frac{V_1}{V_g} = \frac{V_2}{I_1} \cdot \frac{I_1}{V_1} \cdot \frac{V_1}{V_g} = Z_{21} \cdot \frac{1}{Z_{11}} \cdot \frac{Z_{11}}{Z_{11} + R_g} = \frac{Z_{21}}{Z_{11} + R_g} \quad ; \quad \left. R_g \right|_N = 1
$$

$$
\left. \frac{V_2}{V_g} \right|_{I_2=0} = \frac{Z_{21}}{Z_{11} + 1} = K \cdot \frac{s\left(s^2 + \frac{1}{16}\right)}{s^3 + 2 s^2 + 2 s + 1} \quad \rightarrow \quad \frac{V_2}{V_g} = \frac{P(s)}{Q(s)} = \frac{MP + NP}{MQ + NQ}
$$

#### Tenemos entonces una relación entre la transferencia pedida y la relación con nuestro cuadripolo. Para asegurarnos que la síntesis del cuadripolo sea no discipativo, me conviene pensar a la transferencia como:

$$
\begin{cases}
P_{par} := \dfrac{MP}{1 + \dfrac{MQ}{NQ}} \cdot \dfrac{1}{NQ} &  \\[3ex]
P_{impar} := \dfrac{NP}{1 + \dfrac{NQ}{MQ}} \cdot \dfrac{1}{MQ} & 
\end{cases}
$$

#### Analizo denominador:
$$
s^3 + 2 s^2 + 2 s + 1 = MQ + NQ
$$
$$
\frac{NQ}{MQ} = \frac{s^3 + 2 s}{2 s^2 + 1}
$$
#### Analizo numerador:
$$
s\left(s^2 + \frac{1}{16}\right) = s^3 + \frac{1}{16}\, s = MP + NP
$$
$$
\frac{NP}{\text{MP}} = \frac{s^3 + \frac{1}{16}\, s}{1} \quad \rightarrow \quad \text{Falta } MP
$$
$$
\rightarrow \quad \frac{V_2}{V_g} = \frac{P}{Q} = \frac{NP}{1 + \frac{NQ}{MQ}} \cdot \frac{1}{MQ} = \frac{\frac{NP}{MQ}}{1 + \frac{NQ}{MQ}} = \frac{Z_{21}}{Z_{11} + 1}
$$

#### Deducimos entonces en base a la igualación que:
$$
Z_{21} = \frac{s^3 + \frac{1}{16}\, s}{2 s^2 + 1} \quad ; \quad Z_{11} = \frac{s^3 + 2 s}{2 s^2 + 1}
$$

### Síntesis

$$
\left. \frac{V_2}{V_g} \right|_{I_2=0} \quad \rightarrow \quad \text{termina abierto} \quad \rightarrow \quad \text{último componente en derivación}
$$

#### $ \text{Me fijan una corriente de entrada} \rightarrow \text{arranca en derivación.}$ 
Tengo entonces fuente $V_g$, $R_g$ e $I_1$; Planteo la síntesis como $L_1$ en derivación, un tanque serie ($L_2 \parallel C_2$), $L_3$ en derivación y salida abierta con $I_2 = 0$.

*Mapa de polos y ceros mental:*

| Eje | Descripción |
|---|---|
| $T(s)$ | ceros en $0$ y $\frac{1}{4}$ que tengo que respetar |
| $Z_{11}$ | cero en $0$, polo en $\frac{1}{\sqrt{2}}$, cero en $\sqrt{2}$, polo en $\infty$ |
| $\frac{1}{Z_{11}}$ | polo en $0$, cero en $\frac{1}{\sqrt{2}}$, polo en $\sqrt{2}$, cero en $\infty$; $\rightarrow$ remoción parcial en 0 |
| $Y_2$ | polo en $0$, cero ajustado ahora en $\frac{1}{4}$, polo en $\sqrt{2}$, cero en $\infty$; |
| $Z_2$ | cero en $0$, polo en $\frac{1}{4}$, cero en $\sqrt{2}$, polo en $\infty$; $\rightarrow$ Remuevo polo central como tanque serie |
| $Z_4$ | cero en $0$ y polo en $\infty$ |
| $Y_4$ | polo en $0$ y cero en $\infty$; $\rightarrow$ remuevo polo en 0 |

$$
\frac{1}{Z_{11}} = \frac{2 s^2 + 1}{s^3 + 2 s} \quad \rightarrow \quad \text{Remoción parcial } L: \quad \text{recordemos: }Y(s) = \frac{K_0}{s} + K_{\infty}\, s + \sum \frac{2 k_i\, s}{s^2 + {\omega_i}^2}
$$
Tendremos un inductor en derivación.

$$
Y(s) = Y_L = \frac{1}{s L} \quad \rightarrow \quad L = \frac{1}{K_0}
$$
$$
0 = \left. \frac{2 s^2 + 1}{s^3 + 2 s} - \frac{K_0}{s} \right|_{s^2 = -\frac{1}{16}}
$$
$$
K_0 = \left. \frac{1}{Z_{11}} \cdot s \right|_{s^2 = -\frac{1}{16}} = \left. \frac{2 s^2 + 1}{s^3 + 2 s} \cdot s \right|_{s^2 = -\frac{1}{16}} = \left. \frac{2 s^2 + 1}{s^2 + 2} \right|_{s^2 = -\frac{1}{16}} = \frac{14}{31}
$$
$$
L_1 = \frac{1}{K_0} = \frac{31}{14}
$$

#### El resto de la admitancia teniendo en cuenta la remoción parcial nos queda:
$$
Y_2 = \frac{1}{Z_{11}} - \frac{K_0}{s} = \frac{2 s^2 + 1}{s^3 + 2 s} - \frac{\frac{14}{31}}{s} = \frac{2 s^2 + 1 - \left(\frac{14}{31}\, s^2 + 2 \cdot \frac{14}{31}\right)}{s \left(s^2 + 2\right)} = \frac{\frac{48}{31}\, s^2 + \frac{3}{31}}{s \left(s^2 + 2\right)} = \frac{48}{31} \cdot \frac{s^2 + \frac{1}{16}}{s \left(s^2 + 2\right)}
$$

$$
Z_2 = \frac{s \left(s^2 + 2\right)}{\frac{48}{31}\, s^2 + \frac{3}{31}} \quad ; \quad \rightarrow \text{Remoción tanque en } \left(s^2 + \frac{1}{16}\right)
$$
Tanque serie, con la bobina $L_2 $ en paralelo con el capacitor $C_2 $.

$$
2 k_1 = \left. Z_2 \cdot \frac{s^2 + \frac{1}{16}}{s} \right|_{s^2 = -\frac{1}{16}} = \frac{s^2 + 2}{\frac{48}{31}} = \frac{961}{768}
$$

#### Recordemos las 2 inmitancias de $\frac{2 \ k_i \ s}{s²+{w_i}²}$:
$$
\begin{cases}
\dfrac{s}{2 k_i} \\[2ex]
\dfrac{1}{s \cdot \dfrac{2 k_i}{{\omega_i}^2}}
\end{cases}
$$

$$
C_2 = \frac{1}{2 k_1} = \frac{768}{961} \qquad ; \qquad L_2 = \frac{2 k_1}{{\omega_1}^2} = \frac{961}{48}
$$

$$
Z_4 = Z_2 - \frac{2 k_1\, s}{s^2 + \frac{1}{16}} = \frac{s \left(s^2 + 2\right)}{\frac{48}{31}\left(s^2 + \frac{1}{16}\right)} - \frac{\frac{961}{768}\, s}{s^2 + \frac{1}{16}} = \frac{s \left(s^2 + 2\right) - \left(\frac{48}{31} \cdot \frac{961}{768} \cdot s\right)}{\frac{48}{31}\left(s^2 + \frac{1}{16}\right)} = \frac{s^3 + \frac{1}{16}\, s}{\frac{48}{31}\left(s^2 + \frac{1}{16}\right)} = \frac{s}{\frac{48}{31}}
$$

$$
Y_4 = \frac{\frac{48}{31}}{s} \quad \rightarrow \quad \text{Remoción en 0 derivación:} \quad K_0 = \left. Y_4 \cdot s \right| = \frac{48}{31} \quad ; \quad L_3 = \frac{31}{48}
$$

## Síntesis final

| $L_2$ | $\frac{961}{48}$ |

### Verificación:
#### Resolviendo el circuito la transferencia nos queda (asumo $Z_x$ como cada respectiva sección):
$$
\frac{V_2}{V_g} = \frac{Z_1 \parallel \left(Z_2 + Z_3\right)}{R_g + \left[Z_1 \parallel \left(Z_2 + Z_3\right)\right]} \cdot \frac{Z_3}{Z_2 + Z_3} \quad ; \quad R_g = 1
$$

$$
\frac{Z_3}{Z_2 + Z_3} = \frac{s L_3}{\frac{1}{s C_2 + \frac{1}{s L_2}} + s L_3} = \frac{L_3 \left(s^2 C_2 L_2 + 1\right)}{s^2 C_2 L_2 L_3 + L_2 + L_3}
$$

$$
\frac{Z_1 \parallel \left(Z_2 + Z_3\right)}{1 + \left[Z_1 \parallel \left(Z_2 + Z_3\right)\right]} \quad ; \quad Z_1 \parallel \left(Z_2 + Z_3\right) = \frac{1}{\frac{1}{s L_1} + \frac{1}{\frac{1}{s C_2 + \frac{1}{s L_2}} + s L_3}} \rightarrow reemplazo
$$

$$
\frac{Z_1 \parallel \left(Z_2 + Z_3\right)}{1 + \left[Z_1 \parallel \left(Z_2 + Z_3\right)\right]} \quad = \frac{L_1 \, s \left(C_2 L_2 L_3\, s^2 + L_2 + L_3\right)}{1 + \left(s³C_2L_2L_3L_1 + s²C_2L_1L_2 + s²C_2L_2L_3 + sL_1L_2 + sL_1L_3 + L_1+L_2+L_3 \right)} \quad ; \quad
$$

$$
\frac{V_2}{V_g} = \frac{L_1 L_3 \, s \left(s^2 L_2 C_2 + 1\right)}{s^3 C_2 L_1 L_2 L_3 + s^2 \left[C_2 L_1 L_2 + C_2 L_2 L_3\right] + s \left[L_1 L_2 + L_1 L_3\right] + L_1 + L_2 + L_3}
$$
#### Y de acá nos queda entonces:
$$
K = \frac{{L_1 L_3 L_2 C_2}}{C_2 L_1 L_2 L_3} = 1
$$

### Reemplazando los valores se verifica la transferencia

### Verificación simbólica:

```python
C2 = sp.Symbol('C2')
L1 = sp.Symbol('L1')
L2 = sp.Symbol('L2')
L3 = sp.Symbol('L3')

Z1 = s*L1
Y1 = 1/Z1
Y2 = s*C2 + 1/(s*L2)
Z2 = 1/Y2
Z3 = s*L3

Zp = Z3/(Z2+Z3)
Zs = Z2 + Z3
Ys = 1/Zs
Ya = Y1 + Ys
Za = 1/Ya
Zb = Za/(1+Za)

Zt = Zb*Zp

print("Zp(s) =")
display(Zp)

Zp = sp.simplify(Zp)
Zp = sp.together(Zp)
Zp = sp.factor(Zp)
Zp = sp.simplify(Zp)

print("Zp(s) =")
display(Zp)

print("Zb(s) =")
display(Zb)

Zb = sp.simplify(Zb)
Zb = sp.together(Zb)
Zb = sp.factor(Zb)
Zb = sp.simplify(Zb)

print("Zb(s) =")
display(Zb)

print("Zt(s) =")
display(Zt)

Zt = sp.simplify(Zt)
Zt = sp.together(Zt)
Zt = sp.factor(Zt)
Zt = sp.simplify(Zt)

print("Zt(s) =")
display(Zt)
```

## Verificaciones por simulación:

```python
# Defino la transferencia
num = [1, 0, 1/16, 0]
den = [1, 2, 2, 1]
H = sig.TransferFunction(num, den)

w = np.logspace(-2, 1, 1000)   # de 0.01 a 10 rad/s

w_bode, H_mag, H_phase = sig.bode(H, w=w)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6), sharex=True)
fig.suptitle("Respuesta en Frecuencia (Bode)")

ax1.semilogx(w_bode, H_mag, color="purple", lw=2, label="Magnitud")
ax1.set_ylabel("Módulo (dB)")
ax1.grid(True, which="both")
ax1.legend()

ax2.semilogx(w_bode, H_phase, color="orange", lw=2, label="Fase")
ax2.set_ylabel("Fase (°)")
ax2.set_xlabel("Frecuencia $\omega$ [rad/s]")
ax2.grid(True, which="both")
ax2.legend()

plt.tight_layout()
plt.show()
```

# Ejercicio 2:
## Transimpedancia
### Cuadripolo LC con $I_1$ en la entrada, fuente de corriente, cargado con $R_L$ en la salida

$$
T(s) = \frac{V_2}{I_1} = K \cdot \frac{\left(s^2 + 9\right)}{s^3 + 2 s^2 + 2 s + 1}
$$

$$
Denominador: \ \ s^3 + 2 s^2 + 2 s + 1 = (s+1)\left(s + \frac{1}{2} - j\frac{\sqrt{3}}{2}\right)\left(s + \frac{1}{2} + j\frac{\sqrt{3}}{2}\right) = (s+1)\left(s^2 + s + 1\right)
$$

### Entrada de corriente reestringida por $I_1$, salida del circuito cerrada con RL, me conviene pensar al cuadripolo como Z:

$$
\begin{cases}
V_1 = Z_{11} \cdot I_1 + Z_{12} \cdot I_2 \\
V_2 = Z_{21} \cdot I_1 + Z_{22} \cdot I_2
\end{cases}
$$

#### Donde:
$$
RL \cdot -I_2 = V_2 \ o \ I_2 = - \frac{V_2}{RL}
$$
#### Reemplazando nos queda:
$$
\begin{cases}
V_1 = Z_{11} \cdot I_1 + Z_{12} \cdot - \frac{V_2}{RL} \\
V_2 = Z_{21} \cdot I_1 + Z_{22} \cdot - \frac{V_2}{RL}
\end{cases}
$$
#### Y nos damos cuenta en la segunda ecuación que:
$$
\frac{V_2}{I_1} = \left. \frac{Z_{21}}{1 + \frac{Z_{22}}{RL}}\right|_{RL = 1} = \frac{Z_{21}}{1 + Z_{22}}
$$

#### Observando la transferencia nos damos cuenta que ahora responde a:
$$
P_{par} := \dfrac{MP}{1 + \dfrac{MQ}{NQ}} \cdot \dfrac{1}{NQ}
$$

#### Deducimos entonces en base a la igualación que:
$$
Z_{22} = \frac{2 \ s^2 + 1}{s^3 + 2 \ s} \quad ; \quad Z_{21} = \frac{s^2 + 9}{s^3 + 2 \ s}
$$

### Síntesis

$$
\left. \frac{V_2}{I_1} \right|_{I_2= \frac{-V2}{R_L}} \quad \rightarrow \quad \text{analizo de derecha a izquierda empezando por } Z_{22} \quad \rightarrow \quad \text{último componente en derivación}
$$

#### $ \text{Me fijan una corriente de entrada} \rightarrow \text{arranca en derivación.}$ 
Tengo entonces fuente $I_1$, $R_L$ e $I_2$; Planteo la síntesis como $C_1$ en derivación, un tanque serie ($L_2 \parallel C_2$), $C_3$ en derivación.

*Mapa de polos y ceros mental:*

| Eje | Descripción |
|---|---|
| $T(s)$ | ceros en $3$ y en $\infty$ que tengo que respetar |
| $Z_{22}$ | polo en $0$, cero en $\frac{1}{\sqrt{2}}$, polo en $\sqrt{2}$, cero en $\infty$ |
| $\frac{1}{Z_{22}}$ | cero en $0$, polo en $\frac{1}{\sqrt{2}}$, cero en $\sqrt{2}$, polo en $\infty$; $\rightarrow$ remoción parcial en infinito |
| $Y_2$ | cero en $0$, polo en $\frac{1}{\sqrt{2}}$ ,cero ajustado ahora en $3$, polo en $\infty$; |
| $Z_2$ | polo en $0$, cero en $\frac{1}{\sqrt{2}}$, polo en $3$, cero en $\infty$; $\rightarrow$ Remuevo polo central como tanque serie |
| $Z_4$ | polo en $0$ y cero en $\infty$ |
| $Y_4$ | cero en $0$ y polo en $\infty$; $\rightarrow$ remuevo polo en infinito |

#### Remoción parcial en el infinito:
$$
\frac{1}{Z_{22}} = \frac{s^3 + 2 \ s}{2 \ s^2 + 1} \ ; \ \quad Y_C = K_{\infty} \cdot s
$$
$$
0 = \left. \frac{s^3 + 2 \ s}{2 \ s^2 + 1} - {K_{\infty} \cdot s} \right|_{s^2 = -9}
$$
$$
K_{\infty} = \left. \frac{1}{Z_{22}} \cdot \frac{1}{s} \right|_{s^2 = -9} = \left. \frac{s^2 + 2}{2 \ s^2 + 1} \right|_{s^2 = -9} = \left. \frac{-9 + 2}{2 \cdot -9 + 1} \right. = \frac{7}{17}
$$
$$
C_1 = K_{\infty} = \frac{7}{17}
$$

#### El resto de la admitancia teniendo en cuenta la remoción parcial nos queda:
$$
Y_2 = \frac{1}{Z_{22}} - K_{\infty} \cdot s = \frac{s^3 + 2 \ s}{2 \ s^2 + 1} - \frac{7}{17} \cdot s = \frac{s³+2 \ s - \frac{7}{17} \ s \ \left( 2 \ s² + 1 \right) }{2 \ s^2 + 1} = \frac{\frac{3}{17} \ s³ + \frac{27}{17} s }{2 \ s^2 + 1} = \frac{\frac{3}{17} \ s \left( s² + 9\right)}{2 \left( s^2 + \frac{1}{2} \right)}
$$

$$
Z_2 = \frac{2 \left( s^2 + \frac{1}{2} \right)}{\frac{3}{17} \ s \left( s² + 9\right)} \quad ; \quad \rightarrow \text{Remoción tanque en } \left(s^2 + 9\right)
$$
Tanque serie, con la bobina $L_2 $ en paralelo con el capacitor $C_2 $.

$$
2 k_1 = \left. Z_2 \cdot \frac{s^2 + 9}{s} \right|_{s^2 = -9} = \frac{-9 \cdot 2 + 1}{\frac{3}{17} \cdot -9} = \frac{289}{27}
$$

#### Recordemos las 2 inmitancias de $\frac{2 \ k_i \ s}{s²+{w_i}²}$:
$$
\begin{cases}
\dfrac{s}{2 k_i} \\[2ex]
\dfrac{1}{s \cdot \dfrac{2 k_i}{{\omega_i}^2}}
\end{cases}
$$

$$
C_2 = \frac{1}{2 k_1} = \frac{27}{289} \qquad ; \qquad L_2 = \frac{2 k_1}{{\omega_1}^2} = \frac{289}{243}
$$

$$
Z_4 = Z_2 - \frac{2 k_1\, s}{s^2 + 9} = \frac{2 \left( s^2 + \frac{1}{2} \right)}{\frac{3}{17} \ s \left( s² + 9\right)} - \frac{\frac{289}{27}\, s}{s^2 + 9} = \frac{2 \left(s^2 + \frac{1}{2}\right) - \left(\frac{3}{17} \cdot s \cdot \frac{289}{27} \cdot s\right)}{\frac{3}{17} \cdot s \cdot \left(s^2 + 9\right)} = \frac{\frac{1}{9} s² + 1}{\frac{3}{17} \cdot s \cdot \left(s^2 + 9\right)} = \frac{\frac{1}{9}}{\frac{3}{17} s} = \frac{\frac{17}{27}}{s}
$$

$$
Y_4 = \frac{s}{\frac{17}{27}} \quad \rightarrow \quad \text{Remoción en infinito derivación:} \quad K_{\infty} = \left. Y_4 \cdot \frac{1}{s} \right. = \frac{27}{17} \quad ; \quad C_3 = \frac{27}{17}
$$

## Síntesis final

| Componente | Valor |
|---|---|
| $C_1$ | $\frac{7}{17}$ |
| $L_2$ | $\frac{289}{243}$ |
| $C_2$ | $\frac{27}{289}$ |
| $C_3$ | $\frac{27}{17}$ |

### Verificación:
#### Resolviendo el circuito la transferencia nos queda (asumo $Z_x$ como cada respectiva sección):
$$
\frac{V_2}{V_1} = \frac{Z_1 \parallel R_L}{( Z_1 \parallel R_L ) + Z_2} \quad ; \quad R_L = 1
$$

$$
\frac{Z_1 \parallel R_L}{( Z_1 \parallel R_L ) + Z_2} = \frac{ \frac{\frac{1}{sC_1}}{\frac{1}{sC_1} +1} }{\frac{1}{s C_2 + \frac{1}{s L_2}} + \frac{1}{sC_1 +1 }} = \frac{s² C_2 L_2 + 1}{s² C_2 L_2 + s² C_1 L_2 + s L_2 + 1}
$$

#### Ahora nos queda reemplazar $V_1 = I_1 \cdot Z_T$ y para ello hay que hallar $Z_T$ :

$$
Z_T = Z_3 \parallel \left( Z_2 + (Z_1 \parallel R_L) \right)
$$

$$
Z_T = \frac{s^2 C_2 L_2 + s² C_1 L_2 + s L_2 + 1}{s³ C_1 C_2 L_2 + s³ C_1 C_3 L_2 + s³ C_2 C_3 L_2 + s² C_2 L_2 + s^2 C_3 L_2 + sC_3 + sC_1 + 1}
$$

$$
\frac{V_2}{V_1} = \frac{V_2}{I_1 \cdot Z_T} \rightarrow \frac{V_2}{I_1} = \frac{V_2}{V_1} \cdot Z_T
$$

$$
\frac{V_2}{I_1} = \frac{C_2 L_2 \cdot (s^2 + \frac{1}{C_2 L_2})}{s³ (C_2 C_3 L_2 + C_1 C_2 L_2 + C_1 C_3 L_2) + s² (C_2 L_2 + C_3 L_2) + s(C_1 + C_3) + 1}
$$
#### Y de acá nos queda entonces:
$$
K = \frac{C_2 L_2}{C_2 C_3 L_2 + C_1 C_2 L_2 + C_1 C_3 L_2} = \frac{ \frac{27}{289} \cdot \frac{289}{243} }{ \frac{27}{289} \cdot \frac{27}{17} \cdot \frac{289}{243} + \frac{7}{17} \cdot \frac{27}{289} \cdot \frac{289}{243} + \frac{7}{17} \cdot \frac{27}{17} \cdot \frac{289}{243} } = \frac {1}{9}
$$

### Verificación simbólica:

```python
C2 = sp.Symbol('C2')
C1 = sp.Symbol('C1')
L2 = sp.Symbol('L2')
C3 = sp.Symbol('C3')

Y1 = s*C1
Z1 = 1/Y1
Y2 = s*C2 + 1/(s*L2)
Z2 = 1/Y2
Y3 = s*C3
Z3 = 1/Y3

Yp1 = Y1 + 1
Zp1 = 1/Yp1

V2_V1 = Zp1/(Zp1 + Z2)

print("V2/V1(s) =")
display(V2_V1)

V2_V1 = sp.simplify(V2_V1)
V2_V1 = sp.together(V2_V1)
V2_V1 = sp.factor(V2_V1)
V2_V1 = sp.simplify(V2_V1)

print("V2/V1(s) =")
display(V2_V1)

Zs = Z2 + Zp1
Ys = 1/Zs
YT = Y3 + Ys
ZT = 1/YT

print("ZT(s) =")
display(ZT)

ZT = sp.simplify(ZT)
ZT = sp.together(ZT)
ZT = sp.factor(ZT)
ZT = sp.simplify(ZT)

print("ZT(s) =")
display(ZT)

V2_I1 = V2_V1 * ZT

print("V2_I1(s) =")
display(V2_I1)

V2_I1 = sp.simplify(V2_I1)
V2_I1 = sp.together(V2_I1)
V2_I1 = sp.factor(V2_I1)
V2_I1 = sp.simplify(V2_I1)

print("V2_I1(s) =")
display(V2_I1)
```

# Ejercicio 3:
## Transimpedancia
### Cuadripolo LC con $I_g$ en la entrada, fuente de corriente, cargado con $Y_g$ en la entrada, salida abierta

$$
T(s) = \frac{V_2}{I_g} = K \cdot \frac{s \left(s^2 + \frac{1}{16}\right)}{s^3 + 2 s^2 + 2 s + 1}
$$

$$
Denominador: \ \ s^3 + 2 s^2 + 2 s + 1 = (s+1)\left(s + \frac{1}{2} - j\frac{\sqrt{3}}{2}\right)\left(s + \frac{1}{2} + j\frac{\sqrt{3}}{2}\right) = (s+1)\left(s^2 + s + 1\right)
$$

#### $T(s)$ es la misma que en el ejercicio 1, veremos si podemos ahorrar pasos:

### Entrada de tensión reestringida por $Y_g$, salida del circuito abierta, me conviene pensar al cuadripolo como $Z$:
$$
\begin{cases}
V_1 = Z_{11} \cdot I_1 + Z_{12} \cdot I_2 \\
V_2 = Z_{21} \cdot I_1 + Z_{22} \cdot I_2
\end{cases}
$$

$$
Z_{11} = \left. \frac{V_1}{I_1} \right|_{I_2=0} \quad ; \quad Z_{12} = \left. \frac{V_1}{I_2} \right|_{I_1=0}
$$

$$
Z_{21} = \left. \frac{V_2}{I_1} \right|_{I_2=0} \quad ; \quad Z_{22} = \left. \frac{V_2}{I_2} \right|_{I_1=0}
$$

#### $I_1$ será la corriente entrante al cuadripolo por la izquierda, distinto de $I_g$:

### Si reducimos la expresión a la entrada:
$$
Z_{11} = \left. \frac{V_1}{I_1} \right|_{I_2=0}
$$
Cuadripolo reducido a $Z_{11}$ (salida en circuito abierto, $I_2=0$):

$$
I_g = I_{Y_g} + I_{Z_{11} (\text{sería } I_1)} \rightarrow I_g = V_1 \cdot Y_g + \frac{V_1}{Z_{11}} \rightarrow \frac{V_1}{I_g} = \left.  \frac{1}{Y_g + \frac{1}{Z_{11}}} \right|_{Y_g=1} = \frac{1}{1 + \frac{1}{Z_{11}}}
$$

$$
\left. \frac{V_2}{I_g} \right|_{I_2=0} = \frac{V_2}{I_1} \cdot \left(\frac{V_1}{I_1}\right)^{-1} \cdot \frac{V_1}{I_g} = \frac{V_2}{I_1} \cdot \frac{I_1}{V_1} \cdot \frac{V_1}{I_g} = Z_{21} \cdot \frac{1}{Z_{11}} \cdot \frac{1}{1 + \frac{1}{Z_{11}}} = \frac{Z_{21}}{Z_{11} + 1} \quad ; 
$$

$$
\left. \frac{V_2}{I_g} \right|_{I_2=0} = \frac{Z_{21}}{Z_{11} + 1} = K \cdot \frac{s\left(s^2 + \frac{1}{16}\right)}{s^3 + 2 s^2 + 2 s + 1} \quad \rightarrow \quad \frac{V_2}{I_g} = \frac{P(s)}{Q(s)} = \frac{\frac{NP}{MQ}}{1 + \frac{NQ}{MQ}}
$$

#### Deducimos entonces en base a la igualación que:
$$
Z_{21} = \frac{s^3 + \frac{1}{16}\, s}{2 s^2 + 1} \quad ; \quad Z_{11} = \frac{s^3 + 2 s}{2 s^2 + 1}
$$

### Síntesis

Tengo entonces fuente $I_g$, $R_g$; Planteo la síntesis como $L_1$ en derivación, un tanque serie ($L_2 \parallel C_2$), $L_3$ en derivación y salida abierta con $I_2 = 0$.
